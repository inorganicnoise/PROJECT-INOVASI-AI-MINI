```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
```

UPLOAD DATASET

```python
df = pd.read_csv('Carbon Emission.csv')
```

```python
df.shape
```

```python
df.head()
```

CLEANING DAN PREPROCESSING

```python
# pengecekkan data null dan duplikat
null_counts = df.isnull().sum()[lambda s: s > 0]
print("tidak ada data null" if null_counts.empty else  f" Null: {null_counts}")

dup_count = df.duplicated().sum()
print("tidak ada data duplikat" if dup_count == 0 else f"Duplikat: {dup_count}")
```

```python
# mengisi nilai null dengan 'unknown
df['Vehicle Type'] = df['Vehicle Type'].fillna('Unknown')
```

```python
# pembersihan ulang nilai null
df = df.dropna()
```

```python
# pengecekkan ulang
null_counts = df.isnull().sum()[lambda s: s > 0]
print("tidak ada data null" if null_counts.empty else  f" Null: {null_counts}")

dup_count = df.duplicated().sum()
print("tidak ada data duplikat" if dup_count == 0 else f"Duplikat: {dup_count}")
```

```python
# memastikan semua isi kolom CarbonEmission bertipe numerik
df['CarbonEmission'] = pd.to_numeric(df['CarbonEmission'], errors='coerce')
```

```python
median_emisi = df['CarbonEmission'].median()
df['Status_Emisi'] = (df['CarbonEmission'] > median_emisi).astype(int)
print(df['Status_Emisi'].value_counts())
```

```python
median_emisi
```

EXPLORATORY DATA ANALYSIS

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(x='Status_Emisi', hue='Status_Emisi', data=df, palette='Set2', legend=False)

# Lakukan perulangan untuk menampilkan angka di SEMUA batang
for container in ax.containers:
    ax.bar_label(container)

plt.title('Distribusi Status Emisi (0 = Rendah, 1 = Tinggi)')
plt.show()

urutan_terbang = ['never', 'rarely', 'frequently', 'very frequently']
plt.figure(figsize=(7, 4))
sns.boxplot(x='Frequency of Traveling by Air', y='CarbonEmission',
            hue='Frequency of Traveling by Air', data=df,
            order=urutan_terbang, palette='Greens', legend=False)
plt.title('Emisi Karbon Berdasarkan Frekuensi Naik Pesawat')
plt.show()

kolom_num = ['Vehicle Monthly Distance Km', 'How Many New Clothes Monthly',
             'Waste Bag Weekly Count', 'CarbonEmission']
plt.figure(figsize=(7, 5))
sns.heatmap(df[kolom_num].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap Korelasi Fitur Numerik')
plt.show()
```

FEATURE SELECTION

```python
fitur = ['Transport', 'Vehicle Monthly Distance Km', 'How Many New Clothes Monthly',
         'Frequency of Traveling by Air', 'Vehicle Type', 'Waste Bag Weekly Count']

X = pd.get_dummies(df[fitur], drop_first=True, dtype=int)
Y = df['Status_Emisi']
```

TRAIN - TEST SPLIT

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=1, stratify=Y)
```

PEMODELAN

```python
model_dt = DecisionTreeClassifier(max_depth=3, random_state=1)
model_dt.fit(X_train, Y_train)

model_rf = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=1)
model_rf.fit(X_train, Y_train)
```

EVALUASI MODEL

```python
for nama, model in [('Decision Tree', model_dt), ('Random Forest', model_rf)]:
    akurasi = accuracy_score(Y_test, model.predict(X_test))
    cv = cross_val_score(model, X, Y, cv=5).mean()
    print(f'{nama:15s} | Akurasi uji: {akurasi*100:.2f}% | Rata-rata CV 5-fold: {cv*100:.2f}%')

prediksi_rf = model_rf.predict(X_test)
print(classification_report(Y_test, prediksi_rf, target_names=['Rendah', 'Tinggi']))

ConfusionMatrixDisplay.from_predictions(Y_test, prediksi_rf,
                                        display_labels=['Rendah', 'Tinggi'], cmap='Greens')
plt.title('Confusion Matrix - Random Forest')
plt.show()
```

```python
# metrik lengkap Decision Tree
from sklearn.metrics import classification_report

prediksi_dt = model_dt.predict(X_test)
print("=== CLASSIFICATION REPORT: DECISION TREE ===")
print(classification_report(Y_test, prediksi_dt, target_names=['Rendah', 'Tinggi']))
```

```python
imp = pd.Series(model_rf.feature_importances_, index=X.columns)
imp_asli = imp.groupby(lambda kol: next(f for f in fitur if kol.startswith(f))).sum().sort_values()

plt.figure(figsize=(8, 4))
imp_asli.plot(kind='barh', color='seagreen')
plt.title('Faktor yang Paling Berpengaruh (Random Forest)')
plt.xlabel('Tingkat kepentingan')
plt.tight_layout()
plt.show()

print((imp_asli.sort_values(ascending=False) * 100).round(1))
```

```python
plt.figure(figsize=(18, 8))
plot_tree(model_dt, feature_names=X.columns, class_names=['Rendah', 'Tinggi'],
          filled=True, rounded=True, fontsize=8)
plt.title('Decision Tree (max_depth=3)')
plt.show()
```

```python
jumlah_pohon = 6

fig, axes = plt.subplots(
    nrows=2,
    ncols=3,
    figsize=(24, 14)
)

axes = axes.flatten()

for i in range(jumlah_pohon):
    pohon = model_rf.estimators_[i]

    plot_tree(
        pohon,
        feature_names=X.columns,
        class_names=['Rendah', 'Tinggi'],
        filled=True,
        rounded=True,
        max_depth=3,
        fontsize=7,
        ax=axes[i]
    )

    axes[i].set_title(f'Pohon Keputusan ke-{i + 1}')

plt.suptitle(
    'Visualisasi Random Forest (6 Pohon Pertama)',
    fontsize=18
)

plt.tight_layout()
plt.show()
```

PREDIKSI DAN INTERPRETASI

```python

opsi_terbang = urutan_terbang
opsi_kendaraan = ['petrol', 'diesel', 'lpg', 'hybrid', 'electric']
opsi_transport = ['private', 'public', 'walk/bicycle']

label_transport = {
    'private': 'Kendaraan pribadi',
    'public': 'Transportasi umum',
    'walk/bicycle': 'Jalan kaki / sepeda'
}

label_kendaraan = {
    'petrol': 'Bensin',
    'diesel': 'Solar',
    'lpg': 'LPG',
    'hybrid': 'Hybrid',
    'electric': 'Listrik'
}

label_terbang = {
    'never': 'Tidak pernah',
    'rarely': 'Jarang',
    'frequently': 'Sering',
    'very frequently': 'Sangat sering'
}

jarak_umum = int(
    df.loc[
        df['Transport'] == 'public',
        'Vehicle Monthly Distance Km'
    ].median()
)

median_baju = df['How Many New Clothes Monthly'].median()
median_sampah = df['Waste Bag Weekly Count'].median()


def tanya_angka(teks, minimum=0):
    while True:
        try:
            nilai = float(input(teks))

            if nilai < minimum:
                print(f'Nilai minimal {minimum}, coba lagi.')
                continue

            return nilai

        except ValueError:
            print('Masukkan angka yang valid.')


def tanya_pilihan(teks, opsi, label=None):
    label = label or {}

    print(teks.strip(), flush=True)

    for i, o in enumerate(opsi):
        print(f'  [{i}] {label.get(o, o)}', flush=True)

    while True:
        try:
            idx = int(input('Pilih nomor: '))

            if 0 <= idx < len(opsi):
                return opsi[idx]

            print('Nomor di luar pilihan, coba lagi.', flush=True)

        except ValueError:
            print('Masukkan nomor yang valid.', flush=True)


def probabilitas_tinggi(data):
    # Menghitung probabilitas menggunakan model Random Forest
    baris = pd.get_dummies(
        pd.DataFrame([data]),
        dtype=int
    )

    baris = baris.reindex(
        columns=X.columns,
        fill_value=0
    )

    return model_rf.predict_proba(baris)[0][1]


def buat_rekomendasi(data):
    saran = []

    if data['Frequency of Traveling by Air'] in [
        'frequently', 'very frequently'
    ]:
        saran.append(
            'Kurangi perjalanan udara: pilih pertemuan daring '
            'atau gabungkan beberapa urusan dalam satu perjalanan.'
        )

    if data['Vehicle Type'] in ['petrol', 'diesel', 'lpg']:
        saran.append(
            'Pertimbangkan kendaraan hybrid/listrik '
            'atau berbagi tumpangan.'
        )

    if (
        data['Vehicle Type'] != 'Unknown'
        and data['Vehicle Monthly Distance Km'] > jarak_umum
    ):
        saran.append(
            'Jarak tempuh kendaraan pribadi cukup jauh: '
            'coba alihkan sebagian ke transportasi umum.'
        )

    if data['How Many New Clothes Monthly'] > median_baju:
        saran.append(
            'Kurangi pembelian pakaian baru: '
            'coba barang bekas atau tukar pakaian.'
        )

    if data['Waste Bag Weekly Count'] > median_sampah:
        saran.append(
            'Kurangi sampah rumah tangga dengan '
            'memilah dan mendaur ulang.'
        )

    return saran or [
        'Kebiasaanmu sudah ramah lingkungan, pertahankan!'
    ]


def status_dari(skor):
    return 'TINGGI' if skor > 0.5 else 'RENDAH'


def jalankan_prediksi():
    print('=' * 60)
    print(' PREDIKSI STATUS JEJAK KARBON INDIVIDU')
    print('=' * 60)

    # Meminta input transportasi
    transport = tanya_pilihan(
        'Transportasi utama kamu:',
        opsi_transport,
        label_transport
    )

    # Meminta jenis kendaraan jika menggunakan kendaraan pribadi
    if transport == 'private':
        jenis = tanya_pilihan(
            'Jenis kendaraan pribadi:',
            opsi_kendaraan,
            label_kendaraan
        )
    else:
        jenis = 'Unknown'

    # Meminta data kebiasaan sehari-hari
    data = {
        'Transport': transport,
        'Vehicle Monthly Distance Km': tanya_angka(
            'Jarak tempuh per bulan (km): '
        ),
        'How Many New Clothes Monthly': tanya_angka(
            'Jumlah pakaian baru per bulan: '
        ),
        'Frequency of Traveling by Air': tanya_pilihan(
            'Seberapa sering naik pesawat:',
            opsi_terbang,
            label_terbang
        ),
        'Vehicle Type': jenis,
        'Waste Bag Weekly Count': tanya_angka(
            'Jumlah kantong sampah per minggu: '
        )
    }

    # Menampilkan kembali data yang dimasukkan
    print('\n' + '=' * 60)
    print(' DATA YANG KAMU MASUKKAN')
    print('=' * 60)

    print(
        f"Transportasi utama: "
        f"{label_transport.get(transport, transport)}"
    )

    if jenis != 'Unknown':
        print(
            f"Jenis kendaraan: "
            f"{label_kendaraan.get(jenis, jenis)}"
        )

    print(
        f"Jarak tempuh per bulan: "
        f"{data['Vehicle Monthly Distance Km']:g} km"
    )

    print(
        f"Jumlah pakaian baru per bulan: "
        f"{data['How Many New Clothes Monthly']:g}"
    )

    print(
        f"Frekuensi naik pesawat: "
        f"{label_terbang.get(data['Frequency of Traveling by Air'], data['Frequency of Traveling by Air'])}"
    )

    print(
        f"Jumlah kantong sampah per minggu: "
        f"{data['Waste Bag Weekly Count']:g}"
    )

    # Melakukan prediksi
    skor = probabilitas_tinggi(data)
    status = status_dari(skor)

    keterangan = (
        'kurang ramah lingkungan'
        if status == 'TINGGI'
        else 'ramah lingkungan'
    )

    print('=' * 60)
    print(
        f' HASIL: Jejak karbon kamu diprediksi '
        f'{status} ({keterangan})'
    )
    print(
        ' Catatan: ini perkiraan dari pola data, '
        'bukan pengukuran pasti.'
    )
    print('=' * 60)

    
    
    # Evaluasi kebiasaan dan saran
    print('\nEVALUASI KEBIASAAN DAN SARAN')
    print('=' * 60)

    saran = buat_rekomendasi(data)

    # Pisahkan pesan umum dari saran perbaikan
    pesan_umum = 'Kebiasaanmu sudah ramah lingkungan, pertahankan!'
    saran_perbaikan = [
        item for item in saran if item != pesan_umum
    ]

    if status == 'RENDAH':
        print('Status emisi kamu tergolong RENDAH.')
        print('Pertahankan kebiasaan baik yang sudah kamu lakukan!')

        if saran_perbaikan:
            print('\nHal yang masih bisa ditingkatkan:')
            for item in saran_perbaikan:
                print(f'- {item}')
        else:
            print('\nTidak ada saran khusus berdasarkan aturan saat ini.')
            print('Tetap pertahankan kebiasaan baikmu!')

    else:
        print('Status emisi kamu tergolong TINGGI.')

        if saran_perbaikan:
            print('\nSaran untuk mengurangi jejak karbon:')
            for item in saran_perbaikan:
                print(f'- {item}')
        else:
            print('\nBelum ada saran khusus berdasarkan aturan saat ini.')
            print('Coba evaluasi kembali pola konsumsi sehari-harimu.')



jalankan_prediksi()
```
